```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X = df.drop(60, axis="columns")
y = df[60]
```

```python
X
```

```python
y
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_transformed = le.fit_transform(y)
```

```python
y_transformed
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y_transformed, test_size=0.25)
```

```python
X_train.shape, X_test.shape
```

```python
y_train.shape, y_test.shape
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation="relu"),
    keras.layers.Dense(30, activation="relu"),
    keras.layers.Dense(15, activation="relu"),
    keras.layers.Dense(1, activation="sigmoid")
])

model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

model.fit(X_train, y_train, epochs=100, batch_size=8)
```

```python
model.evaluate(X_test, y_test)
```

```python
y_predicted = model.predict(X_test)
y_predicted = np.round(y_predicted)
```

```python
y_predicted[:10]
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_predicted))
```

**Now using the dropout layer in the model after overfitted it.**

```python
model_new = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation="relu"),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(30, activation="relu"),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(15, activation="relu"),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(1, activation="sigmoid")
])

model_new.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

model_new.fit(X_train, y_train, epochs=100, batch_size=8)
```

```python
model_new.evaluate(X_test, y_test)
```

**Here we can accuracy is increasing over the testing data**

```python
y_predicted = model_new.predict(X_test)
y_predicted = np.round(y_predicted)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_predicted))
```
